```python
# where 
cursor.execute("""
SELECT * FROM customers
WHERE age > 20
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
cursor.execute("""
SELECT * FROM customers
WHERE city = 'pune'
""")

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
# where with and 
cursor.execute("""
SELECT * FROM customers
WHERE age > 20 AND city = 'Pune'
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# where with or 
cursor.execute("""
SELECT * FROM customers
WHERE city = 'Pune' OR city = 'Mumbai'
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# where with between 
cursor.execute("""
SELECT * FROM customers
WHERE age BETWEEN 20 AND 22
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# WHERE with IN
cursor.execute("""
SELECT * FROM customers
WHERE city IN ('Pune', 'Satara', 'Kolhapur')
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# where with like 
cursor.execute("""
SELECT * FROM customers
WHERE customer_name LIKE 'A%'
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# ORDER BY
cursor.execute("""
SELECT * FROM customers
ORDER BY age ASC
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
cursor.execute("""
SELECT * FROM customers
ORDER BY age DESC
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# group by 
cursor.execute("""
SELECT city, COUNT(*) AS total_customers
FROM customers
GROUP BY city
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# GROUP BY + AVG
cursor.execute("""
SELECT city, AVG(age) AS average_age
FROM customers
GROUP BY city
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# GROUP BY + SUM
cursor.execute("""
SELECT customer_id, SUM(total_amount) AS total_spent
FROM orders
GROUP BY customer_id
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

```python
cursor.execute("""
SELECT Customers.customer_name,
        customers.city,
        orders.order_id,
        orders.total_amount
FROM customers
JOIN orders
ON customers.customer_id = orders.customer_id
""")
result = cursor.fetchall()

for row in result:
    print(row)
```
